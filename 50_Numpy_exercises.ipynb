```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create a random 5x5 matrix
random_matrix = np.random.random((5, 5))

# Step 2: Normalize the matrix (Z-score normalization)
normalized_matrix = (random_matrix - np.mean(random_matrix)) / np.std(random_matrix)

# Step 3: Plot the heatmaps
plt.figure(figsize=(10, 4))

# Original matrix
plt.subplot(1, 2, 1)
plt.imshow(random_matrix, cmap='viridis', interpolation='nearest')
plt.title("Original Matrix")
plt.colorbar()

# Normalized matrix
plt.subplot(1, 2, 2)
plt.imshow(normalized_matrix, cmap='coolwarm', interpolation='nearest')
plt.title("Normalized Matrix")
plt.colorbar()

plt.tight_layout()
plt.show()
```

```python
 # How to round away from zero a float array ? (★☆☆)

def round_away_from_zero(arr):
    return np.copysign(np.ceil(np.abs(arr)), arr)

x = np.array([-1.2, -1.7, 0, 1.2, 1.7])
rounded = round_away_from_zero(x)

print(rounded)
```

```python
# How to compute ((A+B)*(-A/2)) in place (without copy)

a = np.ones(3) * 1    # a = [1, 1, 1]
b = np.ones(3) * 2    # b = [2, 2, 2]

np.add(a, b, out=b)       # b = a + b → [3, 3, 3]
np.divide(a, 2, out=a)    # a = a / 2 → [0.5, 0.5, 0.5]
a *= -1                   # a = -a → [-0.5, -0.5, -0.5]
np.multiply(a, b, out=b)  # b = b * a → [-1.5, -1.5, -1.5]

print(b)
```

```python
# Extract the integer part of a random array of positive numbers using 4 different methods (★★☆)

# Random array of positive numbers
arr = np.random.uniform(0, 10, 5)
print("Original array:", arr)

# Method 1: np.floor() → rounds down to nearest integer
int_part_1 = np.floor(arr)

# Method 2: astype(int) → type cast (truncates decimal)
int_part_2 = arr.astype(int)

# Method 3: np.trunc() → removes decimal part, same as int()
int_part_3 = np.trunc(arr)

# Method 4: np.fix() → rounds toward zero (same as trunc for positives)
int_part_4 = np.fix(arr)

print(int_part_1)
print(int_part_2)
print(int_part_3)
print(int_part_4)
```

```python
#  Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)



# Create a 10x2 matrix (10 points with x and y)
cartesian = np.random.randn(10, 2)  # Random points

# Extract x and y columns
x = cartesian[:, 0]
y = cartesian[:, 1]

# Compute r and θ for each point
r = np.sqrt(x**2 + y**2)      # Radius
theta = np.arctan2(y, x)      # Angle (radians)

# Combine r and theta into a (10x2) polar array
polar = np.column_stack((r, theta))

print("Cartesian coordinates:\n", cartesian)
print("\nPolar coordinates (r, θ):\n", polar)
```

```python
# How to find the closest value (to a given scalar) in a vector?



# Vector and target scalar
vector = np.array([1.3, 3.7, 5.9, 8.2, 10.5])
target = 6.0

# Find the closest value
closest = vector[np.abs(vector - target).argmin()]

print("Closest value to", target, "is", closest)
```
